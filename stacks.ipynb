```python
def minsubarrays(arr):
        
        n = len(arr)
        min_sum = 0
        stack = []
        for i in range(n+1):
            current = arr[i] if i < n else float("-inf")

            while stack and arr[stack[-1]] >= current:

                mid = stack.pop()
                left_boundary = -1 if not stack else stack[-1]
                right_boundary = i

                count =  (mid - left_boundary)* (right_boundary-mid)
                min_sum += count * arr[mid]
            
            stack.append(i)
        return min_sum % (10**9 + 7)
```
