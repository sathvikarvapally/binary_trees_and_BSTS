```python
class Solution:
    def aggressiveCows(self, arr, k):
        # code here
        arr.sort()
        def possible(min_distance):
            
            cows=1
            last_position = arr[0]
            
            for position in arr[1:]:
                
                if position - last_position>=min_distance:
                    cows+=1
                    last_position=position
                    
                    if cows>=k:
                        return True
                        
            return False                
            
            
            
        
        low = 1
        high = arr[-1]-arr[0]
        
        answer=0
        
        while low<=high:
            
            mid = (low+high)//2
            
            
            
            if possible(mid):
                ans=mid
                low=mid+1
                
            else:
                high = mid-1
                
                
        return ans 
```
