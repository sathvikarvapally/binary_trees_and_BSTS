```python
# level order traversal

from collections import deque

class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:

        if root is None:
            return []
        

        # use deque so remove the root from start and insert its children from left.

        # pick the root into queue

        queue=deque([root])  
        answer=[]
        
        #when queue is not empty

        while queue:
            level_size=len(queue) # this measures in future how many times i need to run the for loop to 
                                  # capture all my child values

            current_level =[]     # list to append the level nodes

            for i in range(level_size):  
                node = queue.popleft()   # take the node out while taking insert its children from 
                                        # right side only when they are not none.

                current_level.append(node.val) # insert the node value to current level list

                if node.left is not None:    # now our node is not none so, we can insert
                                            # its children to the queue for future processing  
                    queue.append(node.left)

                if node.right is not None:
                    queue.append(node.right)

            answer.append(current_level)

        return answer        
```

```python
class Solution:
    def nodetopath(self,root,target):
        if root is None:
            return None

        if root.val==target:
            return [root.val]

        # here iam in binary tree 

        # so, i need to traverse to left and right 

        left_path=self.nodetopath(root.left,target)

        if left_path:
            
            return [root.val]+left_path

        right_path=self.nodetopath(root.right,target)

        if right_path:
            
            return [root.val]+right_path

        return []
```

```python
class Solution:

    def nodetorootpath(self,root,target,path=list()):

        if root is None:
            return None

        if root.val==target:
            path.append(root.val)
            return True
        lans=self.nodetorootpath(root.left,target,path)
        rans=self.nodetorootpath(root.right,target,path)

        if lans or rans:
            path.append(root.val)
            return True

        return False

    def solver(self,root,target):
        path=[]
        self.nodetorootpath(root,target,path)
        path.reverse()
        return path
```

```python
def lca(root,p,q):

    if root is None:
        return 

    if root.val==p or root.val==q:
        return root

    left=lca(root.left,p,q)
    right=lca(root.right,p,q)

    if left is not None and right is not None:
        return root

    return left or right
```

```python
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        diameter=0
        def solver(node):

            nonlocal diameter

            if node is None:
                return 0

            left_tree=solver(node.left)
            right_tree=solver(node.right)

            diameter = max(diameter,left_tree+right_tree)

            return 1+max(left_tree,right_tree)
        solver(root)
        return diameter    
           
```

```python
class Solution:
    def maxPathSum(self, root: Optional[TreeNode]) -> int:

        self.maxSum = -inf

        def dfs(node):

            if not node:
                return 0

            leftgain = max(dfs(node.left),0)
            rightgain = max(dfs(node.right),0)

            self.maxSum = max(self.maxSum, leftgain+node.val+rightgain)

            return max(leftgain+node.val, rightgain + node.val)


        dfs(root)
        return self.maxSum
```

```python
class Solution:
    def verticalTraversal(self, root: Optional[TreeNode]) -> List[List[int]]:
        if root is None:
            return []

        minCol = 0
        maxCol = 0
        columns = defaultdict(list)

        q = deque([(root, 0)])

        while q:
            size = len(q)
            level = defaultdict(list)

            for i in range(size):
                node, col = q.popleft()

                level[col].append(node.val)

                minCol = min(minCol, col)
                maxCol = max(maxCol, col)

                if node.left:
                    q.append((node.left, col - 1))

                if node.right:
                    q.append((node.right, col + 1))

            # Only sort nodes having same row + same column
            for col in level:
                level[col].sort()
                columns[col].extend(level[col])

        ans = []

        for col in range(minCol, maxCol + 1):
            ans.append(columns[col])

        return ans
```

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def verticalTraversal(self, root: TreeNode | None) -> list[list[int]]:

        if root is None:
            return []

        result = []

        # Store: node, row, column
        q = deque([(root, 0, 0)])

        
        cols = defaultdict(list)

        min_col = 0
        max_col = 0

        while q:
            node, row, col = q.popleft()

            min_col = min(min_col, col)
            max_col = max(max_col, col)

            cols[col].append((row, node.val))

            if node.left:
                q.append((node.left, row + 1, col - 1))

            if node.right:
                q.append((node.right, row + 1, col + 1))

        for col in range(min_col, max_col + 1):

            
            cols[col].sort()

            current_column = []

            for row, value in cols[col]:
                current_column.append(value)

            result.append(current_column)

        return result  
```

```python
class Solution:
    def jobSequencing(self, deadline, profit):
        # code here
        jobs = list(zip(deadline,profit))
        
        jobs.sort(key=lambda x:x[1],reverse=True)
        
        max_interval=max(deadline)
        
        slot=list(range(max_interval+1))
        
        count_jobs=0
        total_profit=0
        
        def find(x):
            
            if slot[x]!=x:
                
                slot[x]=find(slot[x])
                
            return slot[x]        
           
        
        
        for d,p in jobs:
            
            available_slot=find(d)
            
            if available_slot>0:
                
                count_jobs+=1
                total_profit+=p
                
                slot[available_slot]=find(available_slot-1)
                
            
            
            
            
            
        return [count_jobs,total_profit]    
```
